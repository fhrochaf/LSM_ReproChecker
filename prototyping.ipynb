```python
from flow_reproassesslsm_st1.crews.reprochecker_crew.reprochecker_crew import ReproducibilityReport

data_output = reprochecker_crew.check_data_reproducibility().output.pydantic       # DataReproOutput
method_output = reprochecker_crew.check_method_reproducibility().output.pydantic   # MethodReproOutput
avail_output = reprochecker_crew.check_artifact_availability().output.pydantic     # AvailabilityOutput
assessment = result2.pydantic.reproducibility_assessment  # only this comes from report_elaborator

final_report = ReproducibilityReport(
    datasets=data_output.datasets,
    methods=method_output.methods,
    availability=avail_output,
    reproducibility_assessment=assessment,
)
```

```python
try:
    print("Saving report")
    output_dir = Path("output")
    output_dir.mkdir(exist_ok=True)
    output_file = output_dir / f"00_repro_report.json"
    with open(output_file, "w") as f:
        f.write(final_report.model_dump_json(indent=2))
    print(f"Report saved to {output_file}")
except Exception as e:
    print(f"Error saving report: {e}.\nFinal report:\n{final_report}")
```
